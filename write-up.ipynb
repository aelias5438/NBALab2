# First Steps: Preparing the data

The first thing I did was combine the salary data set with the statistics you provided, so every single player now has their salary attached to their name. Once the data set was ready, I began to move onto my analysis. The way I completed this analysis was through an algorithm called kMeans. The goal of this algorithm is to group similiar players together based on their statistcs, so we can provide insights about the skill level of any player. Each cluster sits on a continuom from best players to worst, so to find the best value players we looked at those with the lowest salaries in the clusters with the best players. In order to create these clusters, I utilized a few key metrics. These were Pts per minute, Triple doubles, total rebounds, assists, and minutes played. We manually set the number of clusters of players for the alogirthm to sort, so I started with three to test before iterating and improving.

# Improving The Model

To Check if there was a better model with our data, I looked at an elbow plot and a graph of silhoutte scores. The elbow plot is made up of inertia, or a statistic of how varied our data in the clusters is. A low inertia means that each cluster has all its data concentrated inside of it and far from any other cluster, and as inertia rises that means that clusters are getting closer to each other as well as more spread out. We want a low inertia because that means all players in each cluster are likely incredibly similair. On the plot, we increase the number of clusters until we no longer achieve large decreases in inertia. Then we compare that to our silhoutte scores, which measures the cohesion of the clusters, giving a high number when there is not much overlap and a low number when clusters overlap a ton. We compare our best silhoutte scores and inertia values to find that clustering our data with 6 clusters instead of 3 is the way to go
